##### Conceptos de POO (Programación Orientada a Objetos) en Python

### el propmpt en IA gemini lenguaje Python, crear la clase "animal" con los atributos "nombre", "raza" crear el constructor animal, y las funciones ladrar(), correr(), crear los objetos pitbull, chihuahua, nivel principiante, no es interactivo, proporciona tu los datos

```python
# Definición de la clase
class Animal:
    def __init__(self, nombre, raza):
        # El constructor inicializa los atributos
        self.nombre = nombre
        self.raza = raza

    def ladrar(self):
        # Método para simular un ladrido
        print(f"{self.nombre} dice: ¡Guau, guau!🐱‍🏍")

    def correr(
        self
    ):
        # Método para simular que el animal corre
        print(f"{self.nombre} está corriendo velozmente.🐱‍👤")
    def morder(self):
        print(f"{self.nombre} esta mordiendo a la vecina🐱‍👓")

# --- Creación de objetos (Instancias) ---

# Creamos el objeto 'pitbull'
pitbull = Animal("Rocco", "Pitbull")

# Creamos el objeto 'chihuahua'
chihuahua = Animal("Chispa", "Chihuahua")
# creamos el objeto perro callejero con datos
perrocallejero = Animal("solovino", "chafa")
# --- Uso de los objetos ---

print(f"Datos del primer perro: {pitbull.nombre} ({pitbull.raza})")
pitbull.ladrar()
pitbull.correr()

print("-" * 20) # Separador visual

print(f"Datos del segundo perro: {chihuahua.nombre} ({chihuahua.raza})")
chihuahua.ladrar()
chihuahua.correr()

print("-" * 20) # Separador visual

print(f"Datos del tercer perro : {perrocallejero.nombre} ({perrocallejero.raza})")
perrocallejero.ladrar()
perrocallejero.correr()
perrocallejero.morder()
print("programa creado por Karol Castillo NL = 9")
```
